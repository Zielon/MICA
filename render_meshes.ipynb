```python
import trimesh
import numpy as np
from IPython.display import display
from pathlib import Path
from glob import glob
from tqdm.notebook import tqdm
from configs.config import get_cfg_defaults
from generate_mesh import run_generator

image_path_list = [Path(x) for x in glob("/home/jovyan/host/transfab/images/2026*.[jp][pn]g")] 
print(f'Displaying meshes: {image_path_list}')

# Create the scene
scene = trimesh.Scene()
transform = np.eye(4)
transform_padding = 2

# Get the configuration defaults once
config = get_cfg_defaults()

# Iterate over the input images
for image_path in tqdm(image_path_list):
    if image_path.exists():
        mesh = run_generator(image_path, config)
        scene.add_geometry(mesh, node_name=str(image_path.stem), transform=transform)
        transform[0, 3] = transform[0, 3] + mesh.bounds[0][0] + transform_padding
    else:
        print(f"File {image_path} doesn't exist")

display(scene.show(viewer='jupyter', flags={'wireframe': True}))
```
